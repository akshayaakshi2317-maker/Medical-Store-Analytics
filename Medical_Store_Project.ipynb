```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Medicine_ID": ["M101","M102","M103","M104","M105"],

    "Medicine_Name": [
        "Paracetamol",
        "Crocin",
        "Dolo 650",
        "Ibuprofen",
        "Cetirizine"
    ],

    "Category": [
        "Fever",
        "Fever",
        "Fever",
        "Pain Relief",
        "Allergy"
    ],

    "Stock": [120,80,95,70,60],

    "Price": [25,30,35,40,20],

    "Monthly_Sales": [320,280,350,210,190]
}
df = pd.DataFrame(data)

print(df)
```

```python
plt.bar(df["Medicine_Name"], df["Monthly_Sales"])

plt.title("Medicine Sales Analysis")

plt.xlabel("Medicine Name")

plt.ylabel("Monthly Sales")

plt.show()
```

```python
df.groupby("Category")["Stock"].sum().plot(
    kind="pie",
    autopct='%1.1f%%'
)

plt.title("Medicine Category Distribution")

plt.ylabel("")

plt.show()
```

```python
low_stock = df[df["Stock"] < 70]

print(low_stock)
```
